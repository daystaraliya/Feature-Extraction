## Feature Extraction
### Make a new column for episode count

```python
df.loc[4]['Title'] # episode counts in bracket
```

```python
def extract_episode(txt):
  check = False
  data = ''
  for i in txt:
    if i == ')':
      check = False
      return data
    if i == '(':
      check = True
      continue
    if check == True:
      data = data + i


df['Episode'] = df['Title'].apply(extract_episode)
df['Episode'] = df['Episode'].str.replace(" eps","").astype(int)
df.head()

#This one line will extract episodes from title
# df['Title'].str.extract(r'\((.*?)\)')
# Extracting only numeric value from brackets
# df['Title'].str.extract(r'(\d+)').astype(int)
```

### Make a new column for time stamp

```python
df['Title'][49]
```

```python
def time_stamp_extraction(txt):
  check = False
  data = ""
  for i in range(len(txt)):
    if txt[i] == ')':
      for j in range (i+1 , i+20):
        data = data + txt[j]
      return data

df['Time Stamp'] = df['Title'].apply(time_stamp_extraction)
df.head()
```

### Calculate total months

```python
df.head()
```

```python
from datetime import datetime

def total_month(period):
  start_str , end_str = period.split(' - ')
  date_format = "%b %Y"

  start_date = datetime.strptime(start_str , date_format)
  end_date = datetime.strptime(end_str , date_format)

  months = (end_date.year - start_date.year) * 12 + (end_date.month - start_date.month)

  return months + 1

df['Months'] = df['Time Stamp'].apply(total_month)
df.head()

```
