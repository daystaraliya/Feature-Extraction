```python
def extract_epi(txt):
    check = False
    data = ""
    for i in txt:
        if i == ')':
            check = True
            return data
        if i == '(':
            check = True
            continue
        if check == True:
            data += i
df['Episode'] = df['Title'].apply(extract_epi)
df['Episode'] = df['Episode'].str.replace(" eps","").astype(int)
df.head(3)
```

```python
df.loc[0]['Title']
```

```python
def extract_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1 , i+20):
                data = data + txt[j]
            return data    
df['Time'] = df['Title'].apply(extract_time)
df.head()
        
```

```python
from datetime import datetime
def month(period):
    start_str , end_str = period.split(' - ')
    date_format = '%b %Y'
    start_date = datetime.strptime(start_str , date_format)
    end_date = datetime.strptime(end_str , date_format)
    month = (end_date.year - start_date.year)*12 + (end_date.month - start_date.month)
    return month +1

df['Months'] = df['Time'].apply(month)
df.head()
```
